Kịch bản xe tự hành:
Xe tự hành của bạn sử dụng một thanh ghi 8-bit uint8_t vehicle_control để điều khiển trạng thái các thiết bị ngoại vi theo sơ đồ bit sau:

Bit 0: Đèn pha LED trước (0: Tắt, 1: Bật)

Bit 1: Động cơ bên trái (0: Dừng, 1: Chạy)

Bit 2: Động cơ bên phải (0: Dừng, 1: Chạy)

Bit 3: Còi cảnh báo (0: Tắt, 1: Kêu)

Bit 4, 5, 6, 7: Chưa dùng (giữ nguyên là 0)

Xe cũng có một mảng 5 cảm biến dò line:

uint16_t line_sensors[5] = {150, 200, 890, 220, 180};

(Giá trị cảm biến $> 500$ nghĩa là cảm biến đang nằm trên vạch đen, $\le 500$ là nằm ngoài đường trắng).

Yêu cầu lập trình:

1.Khởi tạo vehicle_control = 0 (toàn bộ xe tắt).

2.Dùng toán tử bitwise để:

+ Bật đèn pha LED (Bit 0).

+ Bật cả 2 động cơ trái và phải (Bit 1 và Bit 2).

+ Đảm bảo còi (Bit 3) vẫn đang tắt.

3.Duyệt qua mảng line_sensors bằng vòng lặp for:

+ Đếm xem có bao nhiêu cảm biến phát hiện vạch đen ($> 500$).

+ In giá trị của từng cảm biến ra màn hình.

4.Kiểm tra điều kiện:

+ Dùng phép toán kiểm tra bit để đọc lại trạng thái Bit 1 và Bit 2 từ biến vehicle_control. Nếu cả hai động cơ đều đang bật (1), in ra thông báo: Trang thai: Xe dang di thang.

5.Sau khi hoàn thành nhiệm vụ, dùng toán tử bitwise để tắt toàn bộ động cơ (Clear Bit 1 và Bit 2) nhưng vẫn giữ nguyên đèn pha LED (Bit 0). In giá trị vehicle_control (dưới dạng số nguyên) để xác nhận.

```python
%%writefile day2_bitwise.c
#include <stdio.h>
#include <stdint.h>

int main(void) {
    uint8_t vehicle_control = 0; // Khởi tạo thanh ghi bằng 0
    uint16_t line_sensors[5] = {150, 200, 890, 220, 180};
    
    // BƯỚC 1: Dùng phép toán bitwise bật Bit 0 (LED), Bit 1 (Động cơ trái), Bit 2 (Động cơ phải)
    // Gợi ý: vehicle_control |= ...
    vehicle_control |= (0b111);
    vehicle_control &= ~(1<<3);

    // BƯỚC 2: Duyệt mảng line_sensors, in giá trị từng cảm biến và đếm số cảm biến trên vạch đen (> 500)
    uint8_t count;
    for (uint8_t i=0; i<5; i++) {
        if (line_sensors[i] > 500){
            count += 1;}
            printf("Giá trị của cảm biến %u là %u\n",(unsigned int)i,(unsigned int)line_sensors[i]);
    }
    printf("Số giá tri cảm biến phát hiện vạch đen là %u\n",(unsigned int)count);
    
    // BƯỚC 3: Dùng toán tử bitwise kiểm tra xem Bit 1 và Bit 2 có đang bật không, nếu có thì in "Trang thai: Xe dang di thang"
    if (vehicle_control & (0b11<<1)){
        printf("Trang thai: xe dang di thang\n");
    }

    // BƯỚC 4: Dùng phép toán bitwise tắt Bit 1 và Bit 2 (giữ nguyên Bit 0)
    // In giá trị số nguyên của vehicle_control ra màn hình
    vehicle_control &= ~(0b11<<1);
    printf("Nhi phan vehicle_control: ");
    for (int8_t i=7;i>=0;i--){
        printf("%d",(vehicle_control >> i)&1);
    }

    return 0;
}
```
