NGÀY 1: Hệ thống kiểu dữ liệu cố định kích thước (stdint.h), Ép kiểu và I/O chuẩn trong C

Đề bài: Lập trình mô đun tính toán thông số cảm biến cho xe tự hành
Xe tự hành của bạn có:Một cảm biến pin trả về giá trị ADC 10-bit dạng uint16_t (từ 0 đến 1023, tương ứng với điện áp $0.0\text{V} \to 5.0\text{V}$).Một biến lưu mức băm xung động cơ pwm_speed thuộc kiểu uint8_t (giá trị đặt là 200).Một biến lưu thời gian xe chạy runtime_ms thuộc kiểu uint32_t (giá trị ví dụ: 12500 ms).Yêu cầu:Khai báo đúng các kiểu dữ liệu từ <stdint.h>.Chuyển đổi giá trị ADC (ví dụ cho trước là 768) sang điện áp thực tế (float) theo công thức:$$\text{Voltage} = \frac{\text{ADC} \times 5.0}{1023}$$Tính tốc độ phần trăm của động cơ (float) theo công thức:$$\text{Speed Percent} = \frac{\text{pwm\_speed} \times 100.0}{255}$$In toàn bộ thông số ra màn hình định dạng: 
$$--- HE THONG XE TU HANH ---
$$Thoi gian hoat dong : 12500 ms
$$Gia tri ADC pin     : 768
$$Dien ap pin         : 3.75 V
$$Cong suat dong co   : 78.43 %

```python
%%writefile day1_sensor.c
#include <stdio.h>
#include <stdint.h>
int main(void) {
    // BƯỚC 1: Khai báo và gán giá trị cụ thể cho biến
    uint32_t runtime_ms = 12500;
    uint16_t adc_pin = 768;
    uint8_t pwm_speed = 200;
    
    // BƯỚC 2: Ép kiểu và tính toán
    float voltage = (float)adc_pin * 5.0f / 1023.0f;
    float speed_percent = (float)pwm_speed * 100.0f / 255.0f;
    
    // BƯỚC 3: In kết quả chuẩn định dạng
    printf("--- HE THONG XE TU HANH ---\n");
    printf("Thoi gian hoat dong : %lu ms\n", (unsigned long)runtime_ms);
    printf("Gia tri ADC pin     : %u\n", adc_pin);
    printf("Dien ap pin         : %.2f V\n", voltage);
    printf("Cong suat dong co   : %.2f %%\n", speed_percent);
    
    return 0;
}
```
